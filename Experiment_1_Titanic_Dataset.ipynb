Phase 2: Environment Setup & Data Loading
• Task 1: Import NumPy, Pandas, Matplotlib.pyplot and Seaborn.
• Task 2: Load train.csv into a Pandas DataFrame using pd.read_csv().

```python
# ==========================================
# Phase 2: Environment Setup & Data Loading
# ==========================================

# Task 1: Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Task 2: Upload the Titanic dataset (train.csv)
from google.colab import files
uploaded = files.upload()

# Load train.csv into a Pandas DataFrame
df = pd.read_csv("Titanic-Dataset (1).csv")

# Display the first five rows to verify the dataset
print("Dataset loaded successfully!\n")
print(df.head())
```

Phase 3: Data Inspection
• Task 3: Display first 10 rows using df.head(10).
• Task 4: Display last 5 rows using df.tail(5).
• Task 5: Check dataset dimensions using df.shape (Expected: 891 x 12).
• Task 6: Display dataset information using df.info().
• Task 7: Generate summary statistics using df.describe(); record mean, minimum and
maximum values of Age and Fare.
• Task 8: Identify missing values using df.isnull().sum().

```python

# Phase 3: Data Inspection

# Task 3: Display the first 10 rows
print("========== First 10 Rows ==========")
print(df.head(10))

# Task 4: Display the last 5 rows
print("\n========== Last 5 Rows ==========")
print(df.tail(5))

# Task 5: Check dataset dimensions
print("\n========== Dataset Dimensions ==========")
print("Rows and Columns:", df.shape)

# Task 6: Display dataset information
print("\n========== Dataset Information ==========")
df.info()

# Task 7: Generate summary statistics
print("\n========== Summary Statistics ==========")
print(df.describe())

# Display Mean, Minimum, and Maximum of Age and Fare
print("\n========== Age Statistics ==========")
print("Mean Age   :", df["Age"].mean())
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())

print("\n========== Fare Statistics ==========")
print("Mean Fare   :", df["Fare"].mean())
print("Minimum Fare:", df["Fare"].min())
print("Maximum Fare:", df["Fare"].max())

# Task 8: Identify missing values
print("\n========== Missing Values ==========")
print(df.isnull().sum())
```

Phase 4: Data Manipulation & Cleaning
• Task 9: Fill missing Age values with the median.
• Task 10: Filter surviving passengers (Survived == 1).
• Task 11: Convert Fare column to a NumPy array.
• Task 12: Calculate maximum and average Fare using np.max() and np.mean()..

```python

# Phase 4: Data Manipulation & Cleaning


# Task 9: Fill missing Age values with the median
median_age = df["Age"].median()
df["Age"] = df["Age"].fillna(median_age)

print("Missing values in Age after filling:")
print(df["Age"].isnull().sum())

# Task 10: Filter surviving passengers (Survived == 1)
survivors = df[df["Survived"] == 1]

print("\nFirst 10 Surviving Passengers:")
print(survivors.head(10))

print("\nTotal Number of Survivors:", len(survivors))

# Task 11: Convert Fare column to a NumPy array
fare_array = df["Fare"].to_numpy()

print("\nFare as NumPy Array:")
print(fare_array)

# Task 12: Calculate maximum and average Fare
max_fare = np.max(fare_array)
avg_fare = np.mean(fare_array)

print("\nMaximum Fare:", max_fare)
print("Average Fare:", avg_fare)
```

Phase 5: Data Visualization
• Task 13: Plot histogram of Age.
• Task 14: Plot bar chart of Survival Count.
• Task 15: Plot scatter graph of Age vs Fare.

```python

# Phase 5: Data Visualization

import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------
# Task 13: Plot Histogram of Age
# ------------------------------------------
plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=20, color="skyblue", edgecolor="black")
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.grid(True)
plt.show()

# ------------------------------------------
# Task 14: Plot Bar Chart of Survival Count
# ------------------------------------------
plt.figure(figsize=(6,5))
sns.countplot(x="Survived", data=df, palette="Set2")
plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()

# ------------------------------------------
# Task 15: Plot Scatter Graph of Age vs Fare
# ------------------------------------------
plt.figure(figsize=(8,5))
plt.scatter(df["Age"], df["Fare"], color="red", alpha=0.6)
plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.grid(True)
plt.show()
```
